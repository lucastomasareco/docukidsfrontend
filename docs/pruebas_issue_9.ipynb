# Informe de Pruebas - Issue #9: Subida de archivos y Responsividad

## 1. Verificación del botón "Elegir archivo"
- Se revisó la interfaz en la versión web para asegurar que no se dupliquen componentes y que se muestre un único botón de selección de archivos.

## 2. Protección de funciones nativas (entorno web)
- Se validó el uso de funciones específicas de dispositivos móviles (como llamadas a ajustes o navegación nativa) mediante la validación del entorno (`Platform.OS`), evitando errores al ejecutarse desde un navegador web de PC.

## 3. Pruebas Responsivas
- **Pantallas grandes (PC):** Se verificó la correcta distribución de los elementos, la visibilidad de los botones y la adaptación de los márgenes en pantallas de escritorio.
- **Dispositivos móviles (Celulares):** Se comprobó el correcto funcionamiento táctil y la visualización correcta de los selectores y formularios.
