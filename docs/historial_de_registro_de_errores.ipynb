## 2. Historial de Registro de Errores – Pruebas Web

### 2.1 Error de Inicialización del Cliente de Supabase
- **Qué se encontró:** Pantalla roja de error no capturado (`Uncaught Error: supabaseUrl is required.`) en `lib/supabase.ts` al levantar la versión web de Expo.
- **Severidad:** **Crítica** (bloqueaba el arranque de la app web y la ejecución de tests).
- **Qué se corrigió:** Se eliminaron las aserciones no nulas (`!`) y se asignaron valores de respaldo (*fallbacks*) en `lib/supabase.ts` para `EXPO_PUBLIC_SUPABASE_URL` y `EXPO_PUBLIC_SUPABASE_ANON_KEY`.
- **Estado:** **Resuelto**.

### 2.2 Rechazo de Conexión en Pruebas (`ERR_CONNECTION_REFUSED`)
- **Qué se encontró:** Playwright no podía conectarse a `http://localhost:8081` en los tres navegadores (Chromium, Firefox y WebKit).
- **Severidad:** **Alta** (impedía la ejecución automatizada de las pruebas).
- **Qué se corrigió:** Se configuró el bloque `webServer` en `playwright.config.ts` apuntando a `npm run web` para automatizar el encendido del servidor.
- **Estado:** **Resuelto**.

### 2.3 Restricción de Ejecución de Scripts en PowerShell
- **Qué se encontró:** Error de seguridad `PSSecurityException` en Windows PowerShell al ejecutar comandos de `npx`.
- **Severidad:** **Media**.
- **Qué se corrigió:** Se estandarizó el uso de la terminal **CMD** (Command Prompt) dentro de Visual Studio Code.
- **Estado:** **Resuelto**.

### 2.4 Puntos Postergados y Justificación
- **A. Cobertura E2E de Flujos Complejos de Autenticación:** Se postergó el *mocking* completo de la autenticación remota para no demorar la entrega de la infraestructura E2E multiplataforma de la Issue #12.
- **B. Depuración de Warnings por Deprecación (`uuid`):** Los avisos de deprecación provienen de dependencias indirectas de Expo SDK y no representan un riesgo de seguridad ni afectan los tests.
